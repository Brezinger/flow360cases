```python
import math

# N1, sea level; AoA is treated as the angle between the freestream and propeller axis.
diameter_m = 2.120
rpm = 2186.0
true_airspeed_knots = 91.6
angle_of_attack_deg = 3.5
delta_isa_c = 10.0
air_density_kg_m3 = 1.1839
advance_ratio_given = 0.61

# ISA sea-level temperature plus DISA; dry-air ideal-gas speed of sound.
temperature_k = 273.15 + 15.0 + delta_isa_c
gamma = 1.4
gas_constant_j_kg_k = 287.05287
speed_of_sound_m_s = math.sqrt(gamma * gas_constant_j_kg_k * temperature_k)

revolutions_per_second = rpm / 60.0
true_airspeed_m_s = true_airspeed_knots * 1852.0 / 3600.0
axial_airspeed_m_s = true_airspeed_m_s * math.cos(math.radians(angle_of_attack_deg))
rotational_tip_speed_m_s = math.pi * diameter_m * revolutions_per_second
helical_tip_speed_m_s = math.hypot(rotational_tip_speed_m_s, axial_airspeed_m_s)

rotational_tip_mach = rotational_tip_speed_m_s / speed_of_sound_m_s
helical_tip_mach = helical_tip_speed_m_s / speed_of_sound_m_s
advance_ratio_calculated = true_airspeed_m_s / (revolutions_per_second * diameter_m)

print(f'Air temperature: {temperature_k:.2f} K; speed of sound: {speed_of_sound_m_s:.2f} m/s')
print(f'Rotational tip speed: {rotational_tip_speed_m_s:.2f} m/s')
print(f'Axial airspeed: {axial_airspeed_m_s:.2f} m/s')
print(f'Helical tip speed: {helical_tip_speed_m_s:.2f} m/s')
print(f'Helical tip Mach: {helical_tip_mach:.3f}')
print(f'Rotation-only tip Mach: {rotational_tip_mach:.3f}')
print(f'Advance ratio: {advance_ratio_calculated:.3f} (given {advance_ratio_given:.2f})')
print('Air density is supplied for the case but is not needed to calculate Mach number.')
```
